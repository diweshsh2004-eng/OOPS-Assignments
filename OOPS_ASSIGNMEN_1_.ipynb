1. Factorial Function
# This function uses a simple loop to calculate the factorial of a given number.

```python
def factor(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

# Example usage:
# print(factor(5))  # Output: 120
```

2. String Membership Check
This function uses the in operator to check for the presence of the character 's'.

```python
def check_string(user_input):
    if 's' in user_input.lower():
        print("The string is containing the letter 's'")
    else:
        print("The string doesn't contain the letter 's'")

# Example usage:
# check_string("Success")
```

Palindrome Check (OOPS Approach)
To handle this with Object-Oriented Programming, we create a class that encapsulates the logic for checking if a string reads the same forward and backward.

```python
class PalindromeChecker:
    def __init__(self, word):
        self.word = word

    def is_palindrome(self):
        # Clean the string: remove spaces and make lowercase for accuracy
        cleaned_word = self.word.replace(" ", "").lower()
        # Compare string with its reverse
        if cleaned_word == cleaned_word[::-1]:
            return True
        return False

    def display_result(self):
        if self.is_palindrome():
            print(f"'{self.word}' is a palindrome.")
        else:
            print(f"'{self.word}' is not a palindrome.")

# Taking user input
user_str = input("Enter a string: ")
checker = PalindromeChecker(user_str)
checker.display_result()
```
